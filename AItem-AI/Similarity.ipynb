```python
from OCC.Extend.DataExchange import read_step_file, write_stl_file
import meshio

def step_to_obj(input_step, output_obj):
    # Charger le fichier STEP
    shape = read_step_file(input_step)

    # Exporter le fichier STL
    stl_output_path = output_obj.replace(".obj", ".stl")
    write_stl_file(shape, stl_output_path)

    # Convertir le fichier STL en fichier OBJ
    mesh = meshio.read(stl_output_path)
    meshio.write(output_obj, mesh)
```

```python
import numpy as np
import open3d as o3d
def load_obj(file_path):
    mesh = o3d.io.read_triangle_mesh(file_path)
    return np.asarray(mesh.vertices), np.asarray(mesh.triangles)

def compute_shape_descriptor(vertices, triangles):
    # Compute shape descriptor using, for example, histogram of normal distributions
    # You can use a more sophisticated approach based on your specific needs
    mesh = o3d.geometry.TriangleMesh(vertices=o3d.utility.Vector3dVector(vertices), triangles=o3d.utility.Vector3iVector(triangles))
    normals = np.asarray(mesh.vertex_normals)
    descriptor = np.histogramdd(normals, bins=16, range=[[-1, 1], [-1, 1], [-1, 1]])[0].flatten()

    return descriptor

def compare_shapes(descriptor1, descriptor2):
    # Compare the shape descriptors (you can use any distance metric)
    distance = np.linalg.norm(descriptor1 - descriptor2)
    
    return distance
```

```python

# Load two OBJ files
file_path1 = "C:/Users/kevin/OneDrive - LECNAM/Bureau/AR00000081182.obj"
file_path2 = "C:/Users/kevin/OneDrive - LECNAM/Bureau/AR00000081182.obj"
# file_path2 = "C:/Users/kevin/OneDrive - LECNAM/Bureau/M3D-199690_--A_CATPART_0.obj"
vertices1, triangles1 = load_obj(file_path1)
vertices2, triangles2 = load_obj(file_path2)

# Compute shape descriptors for each model
descriptor1 = compute_shape_descriptor(vertices1, triangles1)
descriptor2 = compute_shape_descriptor(vertices2, triangles2)

# Compare the shape descriptors
similarity_distance = compare_shapes(descriptor1, descriptor2)

print(f"Similarity distance between the two models: {similarity_distance}")
```

```python
import cadquery as cq

def convert_step_to_stl(step_file, stl_file):
    try:
        # Importer le modèle STEP
        result = cq.importers.importStep(step_file)

        # Exporter le modèle STL
        cq.exporters.export(result, stl_file)

        print(f"Conversion réussie de {step_file} vers {stl_file}")
        return True
    except Exception as e:
        print(f"Erreur lors de la conversion de {step_file} vers {stl_file}: {e}")
        return False
```

```python
import numpy as np
from stl import mesh

def calculate_hausdorff_distance(points1, points2):
    distance_matrix1 = np.min(np.linalg.norm(points1[:, np.newaxis, :] - points2, axis=2), axis=1)
    distance_matrix2 = np.min(np.linalg.norm(points1 - points2[:, np.newaxis, :], axis=2), axis=0)

    return max(np.max(distance_matrix1), np.max(distance_matrix2))

def read_stl(file_path):
    mesh_data = mesh.Mesh.from_file(file_path)
    return mesh_data

def extract_points(mesh_data):
    points = np.concatenate(mesh_data.vectors)
    return points

def compare_meshes(file_path1, file_path2):
    mesh_data1 = read_stl(file_path1)
    mesh_data2 = read_stl(file_path2)

    points1 = extract_points(mesh_data1)
    points2 = extract_points(mesh_data2)

    if not points1.any() or not points2.any():
        print("Erreur d'extraction des points. Les listes de points sont vides.")
        return

    # Calculate the Hausdorff distance between the two meshes
    hausdorff_distance = calculate_hausdorff_distance(points1, points2)

    # Do something with the Hausdorff distance (print, save, etc.)
    print(f"Distance de Hausdorff entre les maillages : {hausdorff_distance}")

# Example usage
file_path1 = "C:/Users/kevin/OneDrive - LECNAM/Bureau/AR00000081181.stl"
file_path2 = "C:/Users/kevin/OneDrive - LECNAM/Bureau/AR00000081182.stl"

compare_meshes(file_path1, file_path2)
```

```python
import numpy as np
import tripy

def calculate_hausdorff_distance(points1, points2):
    distance1 = max([min(np.linalg.norm(p - q) for q in points2) for p in points1])
    distance2 = max([min(np.linalg.norm(p - q) for p in points1) for q in points2])
    return max(distance1, distance2)

def read_stl(file_path):
    with open(file_path, 'r') as f:
        lines = f.readlines()

    # Extraire les coordonnées des points
    points = np.array([[float(coord) for coord in line.split()[1:]] for line in lines if line.lower().startswith("vertex")])

    return points

def triangulate_points(points):
    try:
        triangles = tripy.earclip(points)
        return triangles
    except Exception as e:
        print(f"Erreur de triangulation : {str(e)}")
        return None

def compare_meshes(file_path1, file_path2):
    points1 = read_stl(file_path1)
    points2 = read_stl(file_path2)

    triangles1 = triangulate_points(points1)
    triangles2 = triangulate_points(points2)

    if not triangles1 or not triangles2:
        print("Erreur de triangulation. Les listes de triangles sont vides.")
        return

    hausdorff_distance = calculate_hausdorff_distance(triangles1, triangles2)
    print(f"Distance de Hausdorff entre les maillages : {hausdorff_distance}")

# Exemple d'utilisation
file_path1 = "C:/Users/kevin/OneDrive - LECNAM/CNAM - onedrive/Alstom/AItem-Web/3D/AR00000081181.stp"
# file_path2 = "C:/Users/kevin/OneDrive - LECNAM/CNAM - onedrive/Alstom/AItem-Web/3D/AR00000081182.stp"
file_path2 = "C:/Users/kevin/OneDrive - LECNAM/CNAM - onedrive/Alstom/AItem-AI/3D/M3D-224712_B.stp"

compare_meshes(file_path1, file_path2)
```
